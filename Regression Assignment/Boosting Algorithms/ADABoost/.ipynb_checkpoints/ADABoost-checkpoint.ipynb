```python
#importing the Library
#Pandas - For data manipulation & analysis.
import pandas as pd
```

```python
# Reading the Dataset
dataset = pd.read_csv('insurance_pre.csv')
```

```python
dataset.head()
```

```python
#pandas.get_dummies() is a super handy function in pandas that converts categorical variables into one-hot encoded (binary) variables.
#drop_first=True: Drops the first category to avoid multicollinearity (used in regression).
#Use .astype() to explicitly set the data type of the resulting dummy variables — for example, converting from default uint8 to int, float, or bool
datasets=pd.get_dummies(dataset,drop_first=True).astype(int)
```

```python
datasets.head()
```

```python
datasets.columns
```

```python
indep=datasets[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
dep=datasets[['charges']]
```

```python
#Import the scikit-learn library for training and testing data splitting.
from sklearn.model_selection import train_test_split
#X_train is the input value used for training the Model.
#X_test is the input data used to test the trained model.
#y_train is the output value used for training the Model.
#y_test is the output data used to test the trained model.
#The data split ratio is 70% for training and 30% for testing. If random_state is set to 0, the testing data is selected randomly but reproducibly.
X_train, X_test, y_train, y_test = train_test_split(indep, dep, test_size = 1/3, random_state = 0)
```

```python
import xgboost as xgb
from sklearn.metrics import mean_squared_error
```

```python
# Define the model
model = xgb.XGBRegressor(
    objective='reg:squarederror',  # loss function for regression
    n_estimators=100,              # number of trees
    learning_rate=0.1,             # shrinkage rate
    max_depth=3,                   # maximum depth of trees
    random_state=42
)

# Train the model
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Evaluate performance
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse:.4f}")
```

```python
#Imports the r2_score class from the metrics module of the scikit-learn library.
from sklearn.metrics import r2_score
#To validate the test data using the R² (R-squared) evaluation metric in scikit-learn library
r_score=r2_score(y_test,y_pred)
r_score
```

```python
#pickle module in Python is used to save and load the data
import pickle
# Save the training data to a file
pickle.dump(model,open("XGBoost_model.sav","wb"))
model=pickle.load(open("XGBoost_model.sav","rb"))
#Load the saved model file in read mode to predict the user input.
result=model.predict([[50,31,0,1,0]])
#Print the response / result.
result
```

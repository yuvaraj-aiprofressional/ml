```python
#To install the pandas - python library
#Pandas - For data manipulation & analysis.
import pandas as PD
```

```python
dataSet = PD.read_csv("Salary_Data.csv")
```

```python
dataSet
```

```python
inputVal = dataSet[["YearsExperience"]]
inputVal
```

```python
outputVal = dataSet[["Salary"]]
outputVal
```

```python
#To install the scikit-learn - python library
!pip install scikit-learn
```

```python
#Import the scikit-learn library for training and testing data splitting.
from sklearn.model_selection import train_test_split
#X_train is the input value used for training the Model.
#X_test is the input data used to test the trained model.
#y_train is the output value used for training the Model.
#y_test is the output data used to test the trained model.
#The data split ratio is 70% for training and 30% for testing. If random_state is set to 0, the testing data is selected randomly but reproducibly.
X_train,X_test,y_train,y_test=train_test_split(inputVal,outputVal,test_size=0.30,random_state=0)
```

```python
#Imports the LinearRegression class from the linear_model module of the scikit-learn library.
from sklearn.linear_model import LinearRegression
#It allows you to create and train a linear regression model.
regressor=LinearRegression()
#To train (or fit) a machine-learning model on a given dataset.
regressor.fit(X_train,y_train)
```

```python
#To make predictions on new or unseen data after the model has been trained using the fit() method.
y_pred=regressor.predict(X_test)
```

```python
#Imports the r2_score class from the metrics module of the scikit-learn library.
from sklearn.metrics import r2_score
#To validate the test data using the R² (R-squared) evaluation metric in scikit-learn library
r=r2_score(y_test,y_pred)
r
```

```python
#pickle module in Python is used to save and load the data
import pickle
# Save the training data to a file
pickle.dump(regressor,open("resultModel.sav",'wb'))
```
